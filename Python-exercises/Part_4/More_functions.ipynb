Line

```python
def line(length,string):
    if not string:
        print("*"*length)
    else:
        print(string[0]*length)
if __name__=="__main__":
    line(10,"")
    line(10,"*")
    line(10,"-")
```

# A box of hashes

```python
def box_of_hashes(height):
    i=0
    while i<height:
        line(10,"#")
        i += 1
if __name__=="__main__":
    box_of_hashes(5)
```

```python
def square_of_hashes(height):
    i=0
    while i<height:
        line(height,"#")
        i += 1
if __name__=="__main__":
    square_of_hashes(5)
```

# A square

```python
def square(height,char):
    i=0
    while i<height:
        line(height,char)
        i += 1    
if __name__=="__main__":
    square(5,"*")

```

# A triangle

```python
def triangle(a):
    i=0
    while i<a:
        line(i+1,"#")
        i += 1
if __name__=="__main__":
    triangle(5)
```

# A shape

```python
def line(length, string):
    if not string:
        print("*" * length)
    else:
        print(string[0] * length)

def shape(width, character, height, filler):
    i = 1
    while i <= width:
        line(i, character)
        i += 1

    i = 0
    while i < height:
        line(width, filler)
        i += 1

if __name__ == "__main__":
    shape(5, "X", 3, "o")
```

# A spruce

```python
def in_chuoi(string,length):
    if len(string) >= length:
        return string
    length_diff = length - len(string)

    return f"{' ' * (length_diff // 2)}{string}{' ' * (length_diff // 2 + 1)}"


def spruce(height):
    i = 0
    count = 1
    while i < height:

        print(in_chuoi("*"*count,height*2-1))
        i+=1
        count+=2
    print(in_chuoi("*",height*2-1))

if __name__ == "__main__":
    spruce(5)


```

# The greatest number

```python
def greatest_number(a, b, c):
    if a >= b and a >= c:
        return a
    elif b >= a and b >= c:
        return b
    else:
        return c

if __name__ == "__main__":
    print(greatest_number(5, 4, 8))
```

# Same characters

```python
def same_chars(string, first, second):
    if first < 0 or second < 0 or first >= len(string) or second >= len(string):
        return False
    return string[first] == string[second]

if __name__ == "__main__":
    print(same_chars("program", 1, 4))  # True
    print(same_chars("program", 0, 5))  # False
```

# First, second and last words

```python
def first_word(sentence):
    words = sentence.split()
    return words[0]


def second_word(sentence):
    words = sentence.split()
    return words[1]


def last_word(sentence):
    words = sentence.split()
    return words[-1]
if __name__ == "__main__":
    print(first_word("Hello world!"))  # Output: Hello
    print(second_word("Hello world! hoan"))  # Output: world!
    print(last_word("Hello world! hoan" ))  # Output: world!
```
