# From negative to positive

```python
n=int(input("Enter a number: "))
for i in range(-n,n+1):
    if i==0:
        continue
    print(i)
```

# List of stars

```python
def list_of_stars(lis):
    for i in lis:
        print("*"*i)
lis=[1,2,3,4,5]
list_of_stars(lis)
```

# Palindromes

```python
def Palindrome(word):
    return word==word[::-1]
while True:
    word=input("Please type in a palindrome ")
    if Palindrome(word):
        print(f"{word} is a palindrome!")
        break
    else:
        print(f"that wasn't a palindrome")
```

# The sum of positive numbers

```python
def sum_of_positive(lis):
    total=0
    for i in lis:
        if i>0:
            total+=i
    return total
if __name__ == "__main__":
    lis=[-9,2,3,4,5]
    print(f"The result is {sum_of_positive(lis)}")
```

# Even numbers

```python
def even_numbers(lis):
    even=[]
    for i in lis:
        if i%2==0:
            even.append(i)
    return even
if __name__ == "__main__":
    lis=[1,2,3,4,5]
    print(f"original {lis}")
    print(f"new : {even_numbers(lis)}")
```

# The sum of lists

```python
def list_sum(lis1, lis2):
    total=[]
    for i in range(len(lis1)):
        total.append(lis1[i]+lis2[i])
    return total
if __name__ == "__main__":
    lis1=[1,2,3]
    lis2=[4,5,6]
    print(list_sum(lis1,lis2))
```

# Distinct numbers

```python
def distinct_numbers(lis):
    new_lis=[]
    for i in sorted(lis):
        if i not in new_lis:
            new_lis.append(i)
    return new_lis
if __name__ == "__main__":
    my_list = [3, 2, 2, 1, 3, 3, 1]
    print(distinct_numbers(my_list)) # [1, 2, 3]
```

# The length of the longest in the list

```python
def length_of_longest(lis):
    max=len(lis[0])
    for i in lis:
        
        if len(i)>max:
            max=len(i)
            
    return max
my_list = ["adele", "mark", "dorothy", "tim", "hedy", "richard"]

result = length_of_longest(my_list)
print(result)
```

# The shortest in the list

```python
def shortest(lis):
    min=len(lis[0])
    min_string=lis[0]
    for i in lis:
        
        if len(i)<min:
            min=len(i)
            min_string=i
            
    return min_string
my_list = ["adele", "mark", "dorothy", "tim", "hedy", "richard"]

result = shortest(my_list)
print(result)
```

# All the longest in the list

```python
def all_the_longest(lis):
    result = []
    max_length = 0

    for i in lis:
        if len(i) > max_length:
            max_length = len(i)
            result = [i]
        elif len(i) == max_length:
            result.append(i)

    return result
my_list = ["adele", "mark", "dorothy", "tim", "hedy", "richard"]

result = all_the_longest(my_list)
print(result) # ['dorothy', 'richard']
```
