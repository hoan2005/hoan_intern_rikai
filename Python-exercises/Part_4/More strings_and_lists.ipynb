# Everything reversed

```python
def everything_reversed(lis):
    new_lis=[]
    for i in range(len(lis)):
        new_lis.append(lis[len(lis)-1-i][::-1])
    return new_lis
my_list = ["Hi", "there", "example", "one more"]
new_list = everything_reversed(my_list)
print(new_list)
```

# Most common character

```python
def most_common_character(string):
    char=""
    max=0
    for i in string:
        if char == i:
            continue
        if string.count(i)>max:
            max=string.count(i)
            char=i
    return char

second_string = "exemplaryelementary"
print(most_common_character(second_string))    
```

# No vowels allowed

```python
def no_vowels(string):
    for i in string:
        if i in "ueoai":
            string=string.replace(i,"")
    return string
my_string = "this is an example"
print(no_vowels(my_string))
```

# No shouting allowed

```python
def no_shouting(lis):
    new_lis=[]
    for i in lis:
        if not i.isupper():
            new_lis.append(i)
    return new_lis
my_list = ["ABC", "def", "UPPER", "ANOTHERUPPER", "lower", "another lower", "Capitalized"]
pruned_list = no_shouting(my_list)
print(pruned_list)
```

# Neighbours in a list

```python
def longest_series_of_neighbours(lis):
    count=0
    max=count
    for i in range(1,len(lis)):
        if abs(lis[i]-lis[i-1])==1:
            count+=1
            continue
        if count>max:
            max=count
        count=0
    return max+1
my_list = [1, 2, 5, 7, 6, 5, 6, 3, 4, 1, 0]
print(longest_series_of_neighbours(my_list))

```

```python
def Points_average(point_kt, point_bt):
    if len(point_kt) == 0:
        return 0
    return (sum(point_bt) + sum(point_kt)) / len(point_kt)


def Pass_percentage(sinhviendau, soluongSV):
    if soluongSV == 0:
        return 0
    return (sinhviendau / soluongSV) * 100


def quy_diem_bt(lis_so_bt):
    lis_diem_bt = []
    for i in lis_so_bt:
        lis_diem_bt.append(i // 10)
    return lis_diem_bt


def quy_diem_danh_gia(lis_diem_kt, lis_diem_bt):
    lis = [""] * 6
    count = 0

    for i in range(len(lis_diem_bt)):
        diem = lis_diem_bt[i] + lis_diem_kt[i]

        if lis_diem_kt[i] < 10:
            lis[0] += "*"
        elif diem <= 14:
            lis[0] += "*"
        elif diem <= 17:
            lis[1] += "*"
            count += 1
        elif diem <= 20:
            lis[2] += "*"
            count += 1
        elif diem <= 23:
            lis[3] += "*"
            count += 1
        elif diem <= 27:
            lis[4] += "*"
            count += 1
        else:
            lis[5] += "*"
            count += 1

    return lis, count


point_kt = []
lis_so_bt = []

while True:
    data = input("Exam points and exercises completed: ")

    if data == "":
        break

    a, b = data.split()
    a = int(a)
    b = int(b)

    point_kt.append(a)
    lis_so_bt.append(b)

point_bt = quy_diem_bt(lis_so_bt)
point_danh_gia, sv_do = quy_diem_danh_gia(point_kt, point_bt)

print("Statistics:")
print(f"Points average: {Points_average(point_kt, point_bt):.1f}")
print(f"Pass percentage: {Pass_percentage(sv_do, len(point_kt)):.1f}")
print("Grade distribution:")

for grade in range(5, -1, -1):
    print(f"  {grade}: {point_danh_gia[grade]}")
```
