# Change the value of an item

```python
lis=[1,2,3,4,5]
while True:
    index=int(input(" index "))
    if index== -1:
        break
    new_value=(input(" new value "))
    lis[index]=new_value

print(lis)
    
```

# Add items to a list

```python
item=int(input(" item "))
lis=[]
for i in range(item):
    val=int(input(" value "))
    lis.append(val)
print(lis)
```

# Addition and removal

```python
lis=[]
print("The list is now []")
while True:
    choose=input("a(d)d, (r)emove or e(x)it:")
    if choose =="x":
        print("Bye!")
        break
    if not lis:
        
        if choose=="d":
            lis.append(1)
        elif choose=="r":
            print("The list is now []")
    else:
        if choose=="d":
            lis.append(lis[-1]+1)
        elif choose=="r":
            lis.pop()
    print(f"The list is now {lis}")
```

# Same word twice

```python
lis=[]
count=0
while True:
    word=input("Enter a word: ")
    if word in lis:
        print(f"You typed in {count} different words")
        break
    lis.append(word)
    count += 1
```

# List twice

```python
lis=[]
while True:
    value=int(input(" value "))
    if value == 0:
        print("Bye!")
        break
    lis.append(value)
    print(f"The list now: {lis}")
    
    print(f"The list in order: {sorted(lis)}")
```

# The length of a list

```python
def length(lis):
    return len(lis)
if __name__ == "__main__":
    lis=[1,2,3,4,5]
    print(length(lis))
```

# Arithmetic mean

```python
def mean(lis):
    return sum(lis)/len(lis)
if __name__ == "__main__":
    lis=[1,2,3,4,5]
    print(mean(lis))
```

# The range of a list

```python
def range_of_list(lis):
    return max(lis)-min(lis)
if __name__ == "__main__":
    lis=[1,2,3,4,5]
    print(range_of_list(lis))
```
