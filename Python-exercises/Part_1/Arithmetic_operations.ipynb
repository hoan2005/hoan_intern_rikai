# Sum and mean

```python
# Write your solution here
number1 = int(input("Please type in the first number1: "))
number2 = int(input("Please type in the second number2: "))
number3 = int(input("Please type in the third number3: "))
number4 = int(input("Please type in the third number4: "))
sum_=number1+number2+number3+number4
mean=sum_/4
print(f"The sum of the numbers is {sum_} and the mean is {mean}")
```

# Food expenditure

```python
# Write your solution here
times=int(input("How many times a week do you eat at the student cafeteria?"))
price = float(input("The price of a typical student lunch? "))
groceries= float(input("How much money do you spend on groceries in a week? "))

Weekly= times*price+groceries
Daily=Weekly/7
print("Average food expenditure:")
print(f"Daily: {Daily} euros")
print(f"Weekly: {Weekly} euros")
```
