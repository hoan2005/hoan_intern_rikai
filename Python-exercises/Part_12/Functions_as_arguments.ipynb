# Sort by remaining stock

```python
def get_remaining_stock(item: tuple):
    return item[2]


def sort_by_remaining_stock(items: list):
    return sorted(items, key=get_remaining_stock)
products = [("banana", 5.95, 12), ("apple", 3.95, 3), ("orange", 4.50, 2), ("watermelon", 4.95, 22)]

for product in sort_by_remaining_stock(products):
    print(f"{product[0]} {product[2]} pcs")
```

# Sort by number of seasons

```python
def sort_by_seasons(items: list):
    def get_seasons(show: dict):
        return show["seasons"]

    return sorted(items, key=get_seasons)


if __name__ == "__main__":
    shows = [{ "name": "Dexter", "rating" : 8.6, "seasons":9 }, { "name": "Friends", "rating" : 8.9, "seasons":10 },  { "name": "Simpsons", "rating" : 8.7, "seasons":32 }  ]

    for show in sort_by_seasons(shows):
        print(f"{show['name']} {show['seasons']} seasons")
```

# Sort by ratings

```python
def sort_by_ratings(items: list):
    def get_rating(show: dict):
        return show["rating"]

    return sorted(items, key=get_rating, reverse=True)


if __name__ == "__main__":
    shows = [{ "name": "Dexter", "rating" : 8.6, "seasons":9 }, { "name": "Friends", "rating" : 8.9, "seasons":10 },  { "name": "Simpsons", "rating" : 8.7, "seasons":32 }  ]

    print("Rating according to IMDB")
    for show in sort_by_ratings(shows):
        print(f"{show['name']}  {show['rating']}")

```

# ClimbingRoute

```python
class ClimbingRoute:
    def __init__(self, name: str, length: int, grade: str):
        self.name = name
        self.length = length
        self.grade = grade

    def __str__(self):
        return f"{self.name}, length {self.length} metres, grade {self.grade}"


def sort_by_length(routes: list):
    def get_length(route):
        return route.length

    return sorted(routes, key=get_length, reverse=True)


def sort_by_difficulty(routes: list):
    def get_difficulty(route):
        return (route.grade, route.length)

    return sorted(routes, key=get_difficulty, reverse=True)


if __name__ == "__main__":
    r1 = ClimbingRoute("Edge", 38, "6A+")
    r2 = ClimbingRoute("Smooth operator", 11, "7A")
    r3 = ClimbingRoute("Synchro", 14, "8C+")
    r4 = ClimbingRoute("Small steps", 12, "6A+")

    routes = [r1, r2, r3, r4]

    print("Sort by length:")
    for route in sort_by_length(routes):
        print(route)

    print("\nSort by difficulty:")
    for route in sort_by_difficulty(routes):
        print(route)
```

# Climbing areas

```python



class ClimbingArea:
    def __init__(self, name: str):
        self.name = name
        self._routes = []

    def add_route(self, route: ClimbingRoute):
        self._routes.append(route)

    def routes(self):
        return len(self._routes)

    def hardest_route(self):
        if not self._routes:
            return None

        def get_difficulty(route):
            return (route.grade, route.length)

        return max(self._routes, key=get_difficulty)

    def __str__(self):
        hardest = self.hardest_route()
        if hardest is None:
            return f"{self.name}, {self.routes()} routes, no routes"

        return f"{self.name}, {self.routes()} routes, hardest {hardest.grade}"


def sort_by_number_of_routes(areas: list):
    def get_number_of_routes(area):
        return area.routes()

    return sorted(areas, key=get_number_of_routes)


def sort_by_most_difficult(areas: list):
    def get_difficulty(area):
        hardest = area.hardest_route()
        if hardest is None:
            return ("", 0)
        return (hardest.grade, hardest.length)

    return sorted(areas, key=get_difficulty, reverse=True)


if __name__ == "__main__":
    ca1 = ClimbingArea("Olhava")
    ca1.add_route(ClimbingRoute("Edge", 38, "6A+"))
    ca1.add_route(ClimbingRoute("Great cut", 36, "6B"))
    ca1.add_route(ClimbingRoute("Swedish route", 42, "5+"))

    ca2 = ClimbingArea("Nummi")
    ca2.add_route(ClimbingRoute("Synchro", 14, "8C+"))

    ca3 = ClimbingArea("Nalkkila slab")
    ca3.add_route(ClimbingRoute("Small steps", 12, "6A+"))
    ca3.add_route(ClimbingRoute("Smooth operator", 11, "7A"))
    ca3.add_route(ClimbingRoute("Piggy not likey", 12, "6B+"))
    ca3.add_route(ClimbingRoute("Orchard", 8, "6A"))

    areas = [ca1, ca2, ca3]

    print("Sort by number of routes:")
    for area in sort_by_number_of_routes(areas):
        print(area)

    print("\nSort by most difficult:")
    for area in sort_by_most_difficult(areas):
        print(area)
```

# BallPlayers

```python
class BallPlayer:
    def __init__(self, name, number, goals, assists, minutes):
        self.name = name
        self.number = number
        self.goals = goals
        self.assists = assists
        self.minutes = minutes

    def __str__(self):
        return f"BallPlayer(name={self.name}, number={self.number}, goals={self.goals}, passes={self.assists}, minutes={self.minutes})"


def most_goals(players):
    player = max(players, key=lambda p: p.goals)
    return player.name


def most_points(players):
    player = max(players, key=lambda p: p.goals + p.assists)
    return (player.name, player.number)


def least_minutes(players):
    return min(players, key=lambda p: p.minutes)


if __name__ == "__main__":
    player1 = BallPlayer("Archie Bonkers", 13, 5, 12, 46)
    player2 = BallPlayer("Speedy Tickets", 7, 2, 26, 55)
    player3 = BallPlayer("Cruella De Hill", 9, 1, 32, 26)
    player4 = BallPlayer("Devilled Tasmanian", 12, 1, 11, 41)
    player5 = BallPlayer("Donald Quack", 4, 3, 9, 12)

    team = [player1, player2, player3, player4, player5]

    print(most_goals(team))
    print(most_points(team))
    print(least_minutes(team))
```

# Product search

```python
def search(products: list, criterion: callable):
    result = []

    for product in products:
        if criterion(product):
            result.append(product)

    return result


products = [
    ("banana", 5.95, 12),
    ("apple", 3.95, 3),
    ("orange", 4.50, 2),
    ("watermelon", 4.95, 22),
    ("kale", 0.99, 1)
]

print("Chọn cột muốn lọc:")
print("1. Tên sản phẩm")
print("2. Giá sản phẩm")
print("3. Số lượng sản phẩm")

column = int(input("Nhập lựa chọn (1-3): "))

if column == 1:
    value = input("Nhập tên sản phẩm cần tìm: ").lower()

    result = search(
        products,
        lambda p: value in p[0].lower()
    )

elif column == 2:
    operator = input("Chọn điều kiện (<, >, =): ")
    value = float(input("Nhập giá cần so sánh: "))

    if operator == "<":
        result = search(products, lambda p: p[1] < value)
    elif operator == ">":
        result = search(products, lambda p: p[1] > value)
    elif operator == "=":
        result = search(products, lambda p: p[1] == value)
    else:
        result = []
        print("Điều kiện không hợp lệ!")

elif column == 3:
    operator = input("Chọn điều kiện (<, >, =): ")
    value = int(input("Nhập số lượng cần so sánh: "))

    if operator == "<":
        result = search(products, lambda p: p[2] < value)
    elif operator == ">":
        result = search(products, lambda p: p[2] > value)
    elif operator == "=":
        result = search(products, lambda p: p[2] == value)
    else:
        result = []
        print("Điều kiện không hợp lệ!")

else:
    result = []
    print("Lựa chọn không hợp lệ!")

print("\nKết quả tìm kiếm:")
if result:
    for product in result:
        print(product)
else:
    print("Không tìm thấy sản phẩm phù hợp.")
```
