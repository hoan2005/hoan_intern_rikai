# Regular expressions

```python
import re


def is_dotw(my_string: str):
    return re.search(r"Mon|Tue|Wed|Thu|Fri|Sat|Sun", my_string) is not None



def all_vowels(my_string: str):
    return re.fullmatch(r"[aeiou]+", my_string, re.IGNORECASE) is not None



def time_of_day(my_string: str):
    return re.fullmatch(r"(?:[01]\d|2[0-3]):[0-5]\d:[0-5]\d", my_string) is not None


print(time_of_day("12:43:01"))
print(time_of_day("AB:01:CD"))
print(time_of_day("17:59:59"))
print(time_of_day("33:66:77"))
```

# Hockey statistics

```python

import json


def print_player(player):
    name = player["name"]
    team = player["team"]
    goals = player["goals"]
    assists = player["assists"]
    points = goals + assists

    print(f"{name:21}{team:4}{goals:3} + {assists:2} = {points:3}")


def search_player(players):
    name = input("name: ")

    for player in players:
        if player["name"] == name:
            print()
            print_player(player)
            return

    print("Player not found")


def list_teams(players):
    teams = sorted(set(player["team"] for player in players))

    for team in teams:
        print(team)


def list_countries(players):
    countries = sorted(set(player["nationality"] for player in players))

    for country in countries:
        print(country)


def players_in_team(players):
    team = input("team: ")

    selected = [
        player for player in players
        if player["team"] == team
    ]

    selected.sort(
        key=lambda player: player["goals"] + player["assists"],
        reverse=True
    )

    print()
    for player in selected:
        print_player(player)


def players_from_country(players):
    country = input("country: ")

    selected = [
        player for player in players
        if player["nationality"] == country
    ]

    selected.sort(
        key=lambda player: player["goals"] + player["assists"],
        reverse=True
    )

    print()
    for player in selected:
        print_player(player)


def most_points(players):
    how_many = int(input("how many: "))

    selected = sorted(
        players,
        key=lambda player: (
            player["goals"] + player["assists"],
            player["goals"]
        ),
        reverse=True
    )

    print()
    for player in selected[:how_many]:
        print_player(player)


def most_goals(players):
    how_many = int(input("how many: "))

    selected = sorted(
        players,
        key=lambda player: (
            player["goals"],
            -player["games"]
        ),
        reverse=True
    )

    print()
    for player in selected[:how_many]:
        print_player(player)


def main():
    filename = input("file name: ")

    with open(filename, encoding="utf-8") as file:
        players = json.load(file)

    print(f"read the data of {len(players)} players")
    print()

    while True:
        print("commands:")
        print("0 quit")
        print("1 search for player")
        print("2 teams")
        print("3 countries")
        print("4 players in team")
        print("5 players from country")
        print("6 most points")
        print("7 most goals")
        print()

        command = input("command: ")

        if command == "0":
            break

        elif command == "1":
            search_player(players)

        elif command == "2":
            list_teams(players)

        elif command == "3":
            list_countries(players)

        elif command == "4":
            players_in_team(players)

        elif command == "5":
            players_from_country(players)

        elif command == "6":
            most_points(players)

        elif command == "7":
            most_goals(players)

        print()


if __name__ == "__main__":
    main()
```

```python
import requests
import json

url = "https://api.nhle.com/stats/rest/en/skater/summary"

params = {
    "isAggregate": "false",
    "isGame": "false",
    "limit": -1,
    "cayenneExp": "seasonId=20192020 and gameTypeId=2"
}

response = requests.get(url, params=params)
response.raise_for_status()

data = response.json()["data"]

with open("all.json", "w", encoding="utf-8") as f:
    json.dump(data, f, ensure_ascii=False, indent=4)

print(f"Đã lưu {len(data)} bản ghi vào all.json")
```
