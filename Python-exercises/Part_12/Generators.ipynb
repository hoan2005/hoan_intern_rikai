# Even numbers

```python
def even_numbers(beginning: int, maximum: int):
    for number in range(beginning, maximum + 1):
        if number % 2 == 0:
            yield number
numbers = even_numbers(11, 21)
for number in numbers:
    print(number)
```

# Prime numbers

```python
def prime_numbers():
    number = 2

    while True:
        is_prime = True

        for i in range(2, number):
            if number % i == 0:
                is_prime = False
                break

        if is_prime:
            yield number

        number += 1
numbers = prime_numbers()

for i in range(8):
    print(next(numbers))
```

# Random words

```python
import random

def word_generator(characters: str, length: int, amount: int):
    for _ in range(amount):
        word = ""

        for _ in range(length):
            word += random.choice(characters)

        yield word
wordgen = word_generator("abcdefg", 3, 5)

for word in wordgen:
    print(word)
```
