# Times ten

```python
def times_ten(start_index: int, end_index: int):
    dic={}
    for i in range(start_index,end_index+1):
        dic[i]=i*10
    return dic
d = times_ten(3, 6)
print(d)
```

# Factorials

```python
def giai_thua(n: int):
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result
def factorials(n: int):
    dic={}
    for i in range(1,n+1):
        dic[i]=giai_thua(i)
    return dic
k = factorials(5)
print(k[1])
print(k[3])
print(k[5])
```

# Histogram

```python
def histogram(strings : str):
    dic={}
    for i in strings:
        if i not in dic:
            dic[i]="*"
        else:
            dic[i]+="*"
    for i in dic:
        print(f"{i} {dic[i]}")
        
    
histogram("statistically")
```

# Phone book, version 1

```python

phone_book = {}

while True:
    command = input("command (1 search, 2 add, 3 quit): ")

    if command == "1":
        name = input("name: ")
        if name in phone_book:
            print(phone_book[name])
        else:
            print("no number")

    elif command == "2":
        name = input("name: ")
        number = input("number: ")
        phone_book[name] = number
        print("ok!")

    elif command == "3":
        print("quitting...")
        break
```

# Phone book, version 2

```python

phone_book = {}

while True:
    command = input("command (1 search, 2 add, 3 quit): ")

    if command == "1":
        name = input("name: ")
        if name in phone_book:
            for number in phone_book[name]:
                print(number)
        else:
            print("no number")

    elif command == "2":
        name = input("name: ")
        number = input("number: ")

        if name not in phone_book:
            phone_book[name] = []

        phone_book[name].append(number)
        print("ok!")

    elif command == "3":
        print("quitting...")
        break
```

# Invert a dictionary

```python

def invert(dictionary: dict):
    inverted = {}
    for key, value in dictionary.items():
        inverted[value] = key

    dictionary.clear()
    dictionary.update(inverted)


s = {1: "first", 2: "second", 3: "third", 4: "fourth"}
invert(s)
print(s)
```

# Numbers spelled out

```python

def dict_of_numbers():
    numbers = {}

    ones = {
        0: "zero", 1: "one", 2: "two", 3: "three",
        4: "four", 5: "five", 6: "six", 7: "seven",
        8: "eight", 9: "nine"
    }

    teens = {
        10: "ten", 11: "eleven", 12: "twelve",
        13: "thirteen", 14: "fourteen", 15: "fifteen",
        16: "sixteen", 17: "seventeen", 18: "eighteen",
        19: "nineteen"
    }

    tens = {
        2: "twenty", 3: "thirty", 4: "forty",
        5: "fifty", 6: "sixty", 7: "seventy",
        8: "eighty", 9: "ninety"
    }

    for i in range(100):
        if i < 10:
            numbers[i] = ones[i]
        elif i < 20:
            numbers[i] = teens[i]
        else:
            ten = i // 10
            one = i % 10

            if one == 0:
                numbers[i] = tens[ten]
            else:
                numbers[i] = tens[ten] + "-" + ones[one]

    return numbers


numbers = dict_of_numbers()
print(numbers[2])
print(numbers[11])
print(numbers[45])
print(numbers[99])
print(numbers[0])
```

# Movie database

```python

def add_movie(database: list, name: str, director: str, year: int, runtime: int):
    movie = {
        "name": name,
        "director": director,
        "year": year,
        "runtime": runtime
    }
    database.append(movie)


database = []
add_movie(database, "Gone with the Python", "Victor Pything", 2017, 116)
add_movie(database, "Pythons on a Plane", "Renny Pytholin", 2001, 94)
print(database)
```

# Find movies

```python

def find_movies(database: list, search_term: str):
    result = []
    for movie in database:
        if search_term.lower() in movie["name"].lower():
            result.append(movie)
    return result
database = [{"name": "Gone with the Python", "director": "Victor Pything", "year": 2017, "runtime": 116},
{"name": "Pythons on a Plane", "director": "Renny Pytholin", "year": 2001, "runtime": 94},
{"name": "Dawn of the Dead Programmers", "director": "M. Night Python", "year": 2011, "runtime": 101}]

my_movies = find_movies(database, "python")
print(my_movies)
```
