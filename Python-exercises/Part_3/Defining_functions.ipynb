# Print many times

```python
# Write your solution here

# You can test your function by calling it within the following block
def print_many_times(text,times):
    i=0
    while i<times:
        print(text)
        i+=1
if __name__ == "__main__":
    print_many_times("python", 5)
```

```python
# Write your solution here
# You can test your function by calling it within the following block
def hash_square(length):
    i=0
    while i<length:
        print("#"*length)
        i+=1
if __name__ == "__main__":
    hash_square(5)
```

# Chessboard

```python
# Write your solution here

# Testing the function
def chessboard(number):
    i=0
    while i< number:
        j=0
        while j<number:
            if(i+j)%2==0:
                print("1",end="")
            else:
                print("0",end="")
            j+=1
        
            
        print()
        i+=1


if __name__ == "__main__":
    chessboard(3)
```

# A word squared

```python
# Write your solution here
def squared(string, size):
    i = 0
    while i < size * size:
        print(string[i % len(string)], end="")
        i += 1
        if i % size == 0:
            print()
```
