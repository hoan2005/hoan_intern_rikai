# Multiplication

```python
# Write your solution here
number=int(input("Please type in a number:"))
i=1

while i<number+1:
    j=1
    while j<number+1:
        print(f"{i} x {j} = {i*j}")
        j+=1
    i+=1
```

# Factorial

```python
# Write your solution here


while True:
    number=int(input("Please type in a number:"))
    if number<=0:
        print("Thanks and bye!")
        break
    i=2
    gt=1
    while(i<=number):
        gt*=i
        i+=1
    print(f"The factorial of the number {number} is {gt}")
   
```

# Flip the pairs

```python
# Write your solution here
number=int(input("Please type in a number:"))
i=1
while i<=number:
    if i%2==0:
        print(i-1)
    if i%2==1 :
        if i==number:
            print(i)
        else :
            print(i+1)

    i+=1

```

# Taking turns

```python
# Write your solution here
number=int(input("Please type in a number:"))
left=1
right=number
while left<=right:
    if left== right:
        print(left)
        break
    print(left)
    print(right)
    left+=1
    right-=1



```
