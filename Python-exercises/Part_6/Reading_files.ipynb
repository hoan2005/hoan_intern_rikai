# Largest number

```python
def largest():
    with open("number.txt") as new_file:
        max_num = int(next(new_file))
        for line in new_file:
            number = int(line)
            if number > max_num:
                max_num = number
    return max_num
print(largest())
```

# Fruit market

```python

def read_fruits():
    fruits = {}

    with open("fruit.csv") as file:
        for line in file:
            parts = line.strip().split(";")
            fruit = parts[0]
            price = float(parts[1])

            fruits[fruit] = price

    return fruits
print(read_fruits())
```

# Matrix

```python
def read_matrix():
    matrix = []
    with open("matrix.txt") as file:
        for row in file:
            matrix.append([int(number) for number in row.split(",")])
    return matrix


def matrix_sum():
    matrix = read_matrix()
    total = 0
    for row in matrix:
        total += sum(row)
    return total


def matrix_max():
    matrix = read_matrix()
    return max(max(row) for row in matrix)


def row_sums():
    matrix = read_matrix()
    return [sum(row) for row in matrix]
print(matrix_max())
```

# Course grading, part 1

```python

student_info = input("Student information: ")
exercise_data = input("Exercises completed: ")

students = {}

with open(student_info) as file:
    for line in file:
        parts = line.strip().split(";")
        if parts[0] == "id":
            continue

        student_id = parts[0]
        first = parts[1]
        last = parts[2]
        students[student_id] = [first, last]

with open(exercise_data) as file:
    for line in file:
        parts = line.strip().split(";")
        if parts[0] == "id":
            continue

        student_id = parts[0]
        exercises = [int(x) for x in parts[1:]]
        total = sum(exercises)

        first, last = students[student_id]
        print(first, last, total)
```

# Course grading, part 2

```python
student_info = input("Student information: ")
exercise_data = input("Exercises completed: ")
exam_data = input("Exam points: ")

students = {}

with open(student_info) as file:
    for line in file:
        parts = line.strip().split(";")
        if parts[0] == "id":
            continue

        student_id = parts[0]
        first = parts[1]
        last = parts[2]
        students[student_id] = [first, last]

exercises = {}

with open(exercise_data) as file:
    for line in file:
        parts = line.strip().split(";")
        if parts[0] == "id":
            continue

        student_id = parts[0]
        total = sum(int(x) for x in parts[1:])
        exercises[student_id] = total

exam_points = {}

with open(exam_data) as file:
    for line in file:
        parts = line.strip().split(";")
        if parts[0] == "id":
            continue

        student_id = parts[0]
        total = sum(int(x) for x in parts[1:])
        exam_points[student_id] = total

for student_id, (first, last) in students.items():
    exercise_point = exercises[student_id] * 10 // 40
    total_points = exercise_point + exam_points[student_id]

    if total_points <= 14:
        grade = 0
    elif total_points <= 17:
        grade = 1
    elif total_points <= 20:
        grade = 2
    elif total_points <= 23:
        grade = 3
    elif total_points <= 27:
        grade = 4
    else:
        grade = 5

    print(first, last, grade)
```

# Course grading, part 3

```python
students = {}
exercise_points = {}
exam_points = {}

with open("students1.csv") as file:
    next(file)  
    for line in file:
        parts = line.strip().split(";")
        students[parts[0]] = f"{parts[1]} {parts[2]}"

with open("exercises1.csv") as file:
    next(file)  
    for line in file:
        parts = line.strip().split(";")
        exercise_points[parts[0]] = sum(map(int, parts[1:]))

with open("exam_points1.csv") as file:
    next(file)  
    for line in file:
        parts = line.strip().split(";")
        exam_points[parts[0]] = sum(map(int, parts[1:]))

print("Student information: students1.csv")
print("Exercises completed: exercises1.csv")
print("Exam points: exam_points1.csv")
print()

print(f"{'name':30}{'exec_nbr':10}{'exec_pts.':10}"
      f"{'exm_pts.':10}{'tot_pts.':10}{'grade':10}")

for student_id, name in students.items():
    exec_nbr = exercise_points[student_id]
    exec_pts = exec_nbr // 4
    exm_pts = exam_points[student_id]
    tot_pts = exec_pts + exm_pts

    if tot_pts < 15:
        grade = 0
    elif tot_pts < 18:
        grade = 1
    elif tot_pts < 21:
        grade = 2
    elif tot_pts < 24:
        grade = 3
    elif tot_pts < 28:
        grade = 4
    else:
        grade = 5

    print(f"{name:<30}{exec_nbr:<10}{exec_pts:<10}"
          f"{exm_pts:<10}{tot_pts:<10}{grade:<10}")
```

# Spell checker

```python

wordlist = []

with open("wordlist.txt") as file:
    for line in file:
        wordlist.append(line.strip().lower())

text = input("Write text: ")
words = text.split()

result = []

for word in words:
    if word.lower() not in wordlist:
        result.append(f"*{word}*")
    else:
        result.append(word)

print(" ".join(result))
```

# Recipe search

```python

def search_by_name(filename: str, word: str):
    with open(filename, "r") as file:
        content = file.read()

    recipes = content.split("\n\n")
    found_recipes = []

    for recipe in recipes:
        lines = recipe.splitlines()
        name = lines[0]

        if word.lower() in name.lower():
            found_recipes.append(name)

    return found_recipes

def search_by_time(filename: str, prep_time: int):
    file = open(filename)
    content = file.read()
    file.close()

    recipes = content.split("\n\n")
    found_recipes = []

    for recipe in recipes:
        lines = recipe.splitlines()

        name = lines[0]
        time = int(lines[1])

        if time <= prep_time:
            found_recipes.append(
                f"{name}, preparation time {time} min"
            )

    return found_recipes



def search_by_ingredient(filename: str, ingredient: str):
    file = open(filename)
    content = file.read()
    file.close()

    recipes = content.split("\n\n")
    found_recipes = []

    for recipe in recipes:
        lines = recipe.splitlines()

        name = lines[0]
        time = int(lines[1])
        ingredients = lines[2:]

        for item in ingredients:
            if ingredient.lower() == item.lower():
                found_recipes.append(
                    f"{name}, preparation time {time} min"
                )
                break

    return found_recipes

print(search_by_name("recipes1.txt", "cake"))
print(search_by_time("recipes1.txt", 60))
print(search_by_ingredient("recipes1.txt", "salt"))
```

# City bikes

```python
import math
def get_station_data(filename: str):
    with open(filename) as file:
        content = file.read()

    lines = content.splitlines()
    stations = {}

    for line in lines[1:]:
        parts = line.split(";")

        longitude = float(parts[0])
        latitude = float(parts[1])
        name = parts[3]

        stations[name] = (longitude, latitude)

    return stations


def distance(stations: dict, station1: str, station2: str):
    longitude1, latitude1 = stations[station1]
    longitude2, latitude2 = stations[station2]

    x_km = (longitude1 - longitude2) * 55.26
    y_km = (latitude1 - latitude2) * 111.2

    distance_km = math.sqrt(x_km**2 + y_km**2)

    return distance_km
def greatest_distance(stations: dict):
    greatest = 0
    station_names = list(stations.keys())
    result = ()

    for i in range(len(station_names)):
        for j in range(i + 1, len(station_names)):
            station1 = station_names[i]
            station2 = station_names[j]

            d = distance(stations, station1, station2)

            if d > greatest:
                greatest = d
                result = (station1, station2, d)

    return result
stations = get_station_data('Helsinki.csv')
print(stations)
station1, station2, greatest = greatest_distance(stations)
print(station1, station2, greatest)
```
