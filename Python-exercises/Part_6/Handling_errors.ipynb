# Reading input

```python
def read_input(prompt, min_value, max_value):
    while True:
        try:
            number = int(input(prompt))

            if min_value <= number <= max_value:
                return number
            else:
                print(f"You must type in an integer between {min_value} and {max_value}")

        except ValueError:
            print(f"You must type in an integer between {min_value} and {max_value}")


number = read_input("Please type in a number: ", 5, 10)
print("You typed in:", number)
```

# Parameter validation

```python

def new_person(name: str, age: int):
    try:
        if name == "":
            raise ValueError("Name cannot be empty")

        if len(name.split()) < 2:
            raise ValueError("Name must contain at least two words")

        if len(name) > 40:
            raise ValueError("Name is longer than 40 characters")

        if age < 0 or age > 150:
            raise ValueError("Age must be between 0 and 150")

        return (name, age)

    except ValueError as e:
        print("Error:", e)
        raise


try:
    person = new_person("Adn hoan", 20)
    print(person)
    person1 = new_person("Adn", 20)
    
except ValueError:
    print("Cannot create person!")
```

# Incorrect lottery numbers

```python
def filter_incorrect():
    with open("lottery_numbers.csv", "r") as file:
        with open("correct_numbers.csv", "w") as new_file:
            for line in file:
                try:
                    parts = line.strip().split(";")

                    # Check the week format
                    week = parts[0].split()
                    if len(week) != 2 or week[0] != "week":
                        continue

                    week_number = int(week[1])
                    if week_number < 1:
                        continue

                    # Check the lottery numbers
                    numbers = parts[1].split(",")
                    if len(numbers) != 7:
                        continue

                    numbers = [int(n) for n in numbers]

                    if any(n < 1 or n > 39 for n in numbers):
                        continue

                    if len(set(numbers)) != 7:
                        continue

                    # If all checks pass, write the line
                    new_file.write(line)

                except (ValueError, IndexError):
                    continue
filter_incorrect()
```
