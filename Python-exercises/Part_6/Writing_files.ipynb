# Filtering the contents of a file

```python

def filter_solutions():
    with open("solutions.csv") as file:
        lines = file.readlines()

    with open("correct.csv", "a") as correct_file, \
         open("incorrect.csv", "a") as incorrect_file:

        for line in lines:
            parts = line.strip().split(";")

            name = parts[0]
            problem = parts[1]
            result = int(parts[2])

            if "+" in problem:
                operands = problem.split("+")
                answer = int(operands[0]) + int(operands[1])
            else:
                operands = problem.split("-")
                answer = int(operands[0]) - int(operands[1])

            if answer == result:
                correct_file.write(line)
            else:
                incorrect_file.write(line)
(filter_solutions())
```

# Course grading, part 4

```python

students = {}
exercise_points = {}
exam_points = {}

student_file = input("Student information: ")
exercise_file = input("Exercises completed: ")
exam_file = input("Exam points: ")
course_file = input("Course information: ")

with open(student_file) as file:
    next(file)
    for line in file:
        parts = line.strip().split(";")
        students[parts[0]] = f"{parts[1]} {parts[2]}"

with open(exercise_file) as file:
    next(file)
    for line in file:
        parts = line.strip().split(";")
        exercise_points[parts[0]] = sum(map(int, parts[1:]))

with open(exam_file) as file:
    next(file)
    for line in file:
        parts = line.strip().split(";")
        exam_points[parts[0]] = sum(map(int, parts[1:]))

with open(course_file) as file:
    course_name = file.readline().strip().split(": ")[1]
    credits = file.readline().strip().split(": ")[1]

with open("results.txt", "w") as txt_file, \
     open("results.csv", "w") as csv_file:

    txt_file.write(f"{course_name}, {credits} credits\n")
    txt_file.write("=" * 36 + "\n")
    txt_file.write(
        f"{'name':30}{'exec_nbr':10}{'exec_pts.':10}"
        f"{'exm_pts.':10}{'tot_pts.':10}{'grade':10}\n"
    )

    for student_id, name in students.items():
        exec_nbr = exercise_points[student_id]
        exec_pts = exec_nbr // 4
        exm_pts = exam_points[student_id]
        tot_pts = exec_pts + exm_pts

        if tot_pts < 15:
            grade = 0
        elif tot_pts < 18:
            grade = 1
        elif tot_pts < 21:
            grade = 2
        elif tot_pts < 24:
            grade = 3
        elif tot_pts < 28:
            grade = 4
        else:
            grade = 5

        txt_file.write(
            f"{name:<30}{exec_nbr:<10}{exec_pts:<10}"
            f"{exm_pts:<10}{tot_pts:<10}{grade:<10}\n"
        )

        csv_file.write(f"{student_id};{name};{grade}\n")

print("Results written to files results.txt and results.csv")
```

# Word search

```python
def kiem_tra(word_search,word2):
    for i in range(len(word_search)):
        if word_search[i] != "." and word_search[i]!=word2[i]:
            return False
    return True
def find_words(search_term: str):
    with open("words.txt") as file:
        words = file.read().split()

        lis=[]
        for i in words:
            lenght=len(search_term)
            if "*" == search_term[0] :

                if i.endswith(search_term[1:]):
                    
                    lis.append(i)
            elif "*"==search_term[-1]:
                if i.startswith(search_term[:-1]):
                    lis.append(i)
            elif "." in search_term:
                if len(i)==lenght and kiem_tra(search_term,i):
                    lis.append(i)
            elif(search_term==i):
                lis.append(i)
    return lis
                
            
                
print(find_words("*e"))

```

# Dictionary stored in a file

```python

while True:
    print("1 - Add word, 2 - Search, 3 - Quit")
    function = input("Function: ")

    if function == "1":
        finnish = input("The word in Finnish: ")
        english = input("The word in English: ")

        with open("dictionary.txt", "a") as file:
            file.write(f"{finnish} - {english}\n")

        print("Dictionary entry added")

    elif function == "2":
        search_term = input("Search term: ")

        with open("dictionary.txt") as file:
            for line in file:
                parts = line.strip().split(" - ")
                finnish = parts[0]
                english = parts[1]

                if search_term in finnish or search_term in english:
                    print(f"{finnish} - {english}")

    elif function == "3":
        print("Bye!")
        break
```
