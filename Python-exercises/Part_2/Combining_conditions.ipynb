# Age check

```python
# Write your solution here
age=int(input("What is your age?"))
if age<0 or age > 110:
    print("That must be a mistake")
elif age <5 :
    print("I suspect you can't write quite yet...")
else :
    print(f"Ok, you're {age} years old")
```

# Grades and points

```python
# Write your solution here
points=float(input("How many points [0-100]: "))
if 0<=points<=49:
    print("fail")
elif 50<=points<=59:
    print("Grade: 1")
elif 60<=points<=69:
    print("Grade: 2") 
elif 70<=points<=79:
    print("Grade: 3")
elif 80<=points<=89:
    print("Grade: 4")
elif 90<=points<=100:
    print("Grade: 5")
else:
    print("Grade: impossible!")
```

# FizzBuzz

```python
# Write your solution here
number=int(input("number : "))
if number%3==0 and number%5==0 :
    print("FizzBuzz")
elif number%3==0:
    print("Fizz")
elif number%5==0:
    print("Buzz")

```

# Leap year

```python
# Write your solution here
year=int(input("Please type in a year: "))
if year%400==0 or (year%100!=0 and year%4==0):
   
    print("That year is a leap year")
else:
    print("That year is not a leap year")
    
```

# Alphabetically in the middle

```python
# Write your solution here
letter1=input("1st letter:")
letter2=input("2st letter:")
letter3=input("3st letter:")
letter=letter1
if letter1<letter2<letter3 or letter3<letter2<letter1 :
    letter=letter2
elif letter1<letter3<letter2 or letter2<letter3<letter1:
    letter=letter3
print(f"The letter in the middle is {letter}")
```

# Gift tax calculator

```python
# Write your solution here
gift=int(input("Value of gift:"))
if 5000>gift:
    print("No tax!")
elif 5000<=gift<=25000:
    tax=100+(gift-5000)*0.08
    print(f"Amount of tax: {tax} euros")
elif 25000<=gift<=55000:
    tax=1700+(gift-25000)*0.1
    print(f"Amount of tax: {tax} euros")
elif 55000<=gift<=200000:
    tax=4700+(gift-55000)*0.12
    print(f"Amount of tax: {tax} euros")
elif 200000<=gift<=1000000:
    tax=22100+(gift-200000)*0.15
    print(f"Amount of tax: {tax} euros")
else:
    tax=142100+(gift-1000000)*0.17
    print(f"Amount of tax: {tax} euros")
```
