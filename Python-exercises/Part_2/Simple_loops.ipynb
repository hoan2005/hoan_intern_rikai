# Input validation

```python
from math import sqrt
# Write your solution here
while True:
    number=int(input("Please type in a number:"))
    if number==0:
        print("Exiting...")
        break
    if number<0:
        print("Invalid number")
    else:
        print(sqrt(number))
```

# The next leap year

```python

year=int(input("year"))
now_year=year
if year==1:
    print(f"The next leap year after {now_year} is {year+3}")
elif not(year%400==0 or (year%4==0 and year% 100 !=0) or year !=1 ):
    print(f"The next leap year after {now_year} is {year+1}")
else:
    while True :
        year+=1
        if year%400==0 or (year%4==0 and year% 100 !=0):
            print(f"The next leap year after {now_year} is {year}")
            break
```

# Working with numbers

```python
# Write your solution here
count=0
sum_=0
count_Positives=0
count_negatives=0
while True:
    number=int(input("Please type in integer numbers. Type in 0 to finish. "))
    if number==0:
        break
    elif number>0:
        count_Positives+=1
    else:
        count_negatives+=1
    count+=1
    sum_+=number
print("Please type in integer numbers. Type in 0 to finish.")
print(f"Numbers typed in {count}")
print(f"The sum of the numbers is {sum_}")
print(f"The mean of the numbers is {sum_/count}")
print(f"Positive numbers {count_Positives}")
print(f"Negative numbers {count_negatives}")


```
