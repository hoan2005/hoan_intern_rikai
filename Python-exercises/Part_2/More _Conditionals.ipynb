# Greater than or equal to

```python
# Write your solution here
integer_a=int(input("Please type in the first number"))
integer_b=int(input("Please type in another number"))
if integer_a>integer_b:
    print(f"The greater number was: {integer_a}")
elif integer_a<integer_b:
    print(f"The greater number was: {integer_b}")
else :
    print(f"The numbers are equal!")

```

# The elder

```python
# Write your solution here
print("Person 1 :")
name_1=input("Name_1 : ")
age_1=int(input("age_1 : "))
print("Person 2 :")
name_2=input("Name_2 : ")
age_2=int(input("age_2 : "))
if age_1>age_2:
    print(f"The elder is {name_1}")
elif age_1<age_2:
    print(f"The elder is {name_2}")
else :
    print(f"{name_1} and {name_2} are the same age")
```

# Alphabetically last

```python
string_1=(input("Please type in the 1st word: ")).lower()
string_2=(input("Please type in the 2nd word: ")).lower()
if string_1>string_2:
    print(f"{string_1} comes alphabetically last.")
elif string_2>string_1:
    print(f"{string_2} comes alphabetically last.")
else:
    print("You gave the same word twice.")
```
