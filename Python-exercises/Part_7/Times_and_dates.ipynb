# Valid PIC?

```python


def is_it_valid(pic: str):
    if len(pic) != 11:
        return False

    day = int(pic[0:2])
    month = int(pic[2:4])
    year = int(pic[4:6])
    century = pic[6]
    identifier = pic[7:10]
    control = pic[10]

    if century == "+":
        year += 1800
    elif century == "-":
        year += 1900
    elif century == "A":
        year += 2000
    else:
        return False


    try:
        datetime.date(year, month, day)
    except ValueError:
        return False

  
    if not identifier.isdigit():
        return False


    numbers = pic[0:6] + identifier
    remainder = int(numbers) % 31

    control_characters = "0123456789ABCDEFHJKLMNPRSTUVWXY"
    expected_control = control_characters[remainder]

    return control == expected_control


print(is_it_valid("230827-906F"))  
print(is_it_valid("120488+246L"))  
print(is_it_valid("310823A9877"))  
```

# Screen time

```python


filename = input("Filename: ")

starting_date = input("Starting date: ")
days = int(input("How many days: "))


start_date = datetime.datetime.strptime(starting_date, "%d.%m.%Y").date()


screen_times = []
total_minutes = 0

print("Please type in screen time in minutes on each day (TV computer mobile):")

for i in range(days):
    current_date = start_date + datetime.timedelta(days=i)
    date_string = current_date.strftime("%d.%m.%Y")

    data = input(f"Screen time {date_string}: ")
    times = [int(x) for x in data.split()]

    screen_times.append((date_string, times))
    total_minutes += sum(times)

average_minutes = total_minutes / days

end_date = start_date + datetime.timedelta(days=days - 1)

with open(filename, "w") as file:
    file.write(
        f"Time period: {start_date.strftime('%d.%m.%Y')}-"
        f"{end_date.strftime('%d.%m.%Y')}\n"
    )
    file.write(f"Total minutes: {total_minutes}\n")
    file.write(f"Average minutes: {average_minutes}\n")

    for date_string, times in screen_times:
        file.write(f"{date_string}: {'/'.join(map(str, times))}\n")

print(f"Data stored in file {filename}")
```
