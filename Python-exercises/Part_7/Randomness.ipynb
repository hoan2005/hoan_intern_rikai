# Randomness

```python
import random

def lottery_numbers(amount: int, lower: int, upper: int):
    numbers = random.sample(range(lower, upper + 1), amount)
    return sorted(numbers)
for number in lottery_numbers(7, 1, 40):
    print(number)
```

# Password generator, part 1

```python
def generate_password(length: int):
    password = ""
    for i in range(length):
        password += random.choice(string.ascii_lowercase)
    return password
for i in range(10):
    print(generate_password(8))
```

# Password generator, part 2

```python

def generate_strong_password(length: int, numbers: bool, special_characters: bool):
    lowercase = string.ascii_lowercase
    digits = string.digits
    special = "!?=+-()#"
    characters = lowercase
    password = ""
    password += random.choice(lowercase)
    
    if numbers:
        characters += digits
        password += random.choice(digits)

    if special_characters:
        characters += special
        password += random.choice(special)

    while len(password) < length:
        password += random.choice(characters)


    password = "".join(random.sample(password, len(password)))

    return password
for i in range(10):
    print(generate_strong_password(8, True, True))
```

# Dice roller

```python


def roll(die: str):
    dice = {
        "A": [3, 3, 3, 3, 3, 6],
        "B": [2, 2, 2, 5, 5, 5],
        "C": [1, 4, 4, 4, 4, 4]
    }
    return random.choice(dice[die])


def play(die1: str, die2: str, times: int):
    a = 0  # Số lần die1 thắng
    b = 0  # Số lần die2 thắng
    c = 0  # Số lần hòa

    for i in range(times):
        result1 = roll(die1)
        result2 = roll(die2)

        if result1 > result2:
            a += 1
        elif result1 < result2:
            b += 1
        else:
            c += 1

    return (a, b, c)
for i in range(20):
    print(roll("A"), " ", end="")
print()
result = play("A", "C", 1000)
print(result)
result = play("B", "B", 1000)
print(result)
```

```python


def words(n: int, beginning: str):
    matching_words = []

    with open("words.txt", "r") as file:
        for line in file:
            # print(line)
            for word in line.split():
                if word.startswith(beginning):
                    matching_words.append(word)
    

    
    matching_words = list(set(matching_words))
    # print(matching_words)

    try:
        if len(matching_words) < n:
            raise ValueError("Not enough matching words")

    except ValueError as e:
        print(e)
        return []
    return random.sample(matching_words, n)

    

word_list = words(3, "ca")
for word in word_list:
    print(word)
    
```
